### Fixing data types(messy date column)

```python
cust_df= pd.read_csv('step2_categories.csv')
print('sample raw date values:')
print(cust_df['purchase_date'].sample(10, random_state=1).tolist())
print()
print('dtype:', cust_df['purchase_date'].dtype)
```

```python
def parse_date(val):
    val = str(val).strip()
    #ISO format: YYYY-MM-DD
    if len(val) == 10 and val[4] == '-':
        return pd.to_datetime(val, format='%Y-%m-%d')
    elif '-' in val:
        return pd.to_datetime(val, format='%m-%d-%Y')
    elif '/' in val:
        return pd.to_datetime(val, format='%d/%m/%Y')
    return pd.NaT


cust_df['purchase_date'] = cust_df['purchase_date'].apply(parse_date)

print('dtype after converion:', cust_df['purchase_date'].dtype)
print()
print('Sample parsed dates:')
print(cust_df['purchase_date'].sample(8, random_state=1).tolist())
print()
print('Any failed parses (NaT)?', cust_df['purchase_date'].isna().sum())
print()
print('Date range:', cust_df['purchase_date'].min(), 'to', cust_df['purchase_date'].max())

cust_df.to_csv('step3_dates.csv', index=False)
```

### Handling extreme outlier

```python
cust_df = pd.read_csv('step4_invalid.csv')
Q1 = cust_df['purchase_amount'].quantile(0.25)
Q3 = cust_df['purchase_amount'].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

print(f'Q1={Q1:.2f}, Q3={Q3:.2f}, IQR={IQR:.2f}')
print(f'Outlier bounds: [{lower_bound:.2f}, {upper_bound:.2f}]')
print()
outliers = cust_df[cust_df['purchase_amount'] > upper_bound]
print(f'Rows above upper bound: {len(outliers)}')
print(outliers[['order_id', 'product_category', 'purchase_amount']].sort_values('purchase_amount', ascending=False).head(10))
```

##### always check whether outliers should be evaluated within a subgroup, not globally. for instance electronics seems expensive in that category and we risk deleting it as an outlier which would skew our data.

```python
cust_df = pd.read_csv('step4_invalid.csv')

def flag_outliers(group):
    Q1, Q3 = group.quantile(0.25), group.quantile(0.75)
    IQR = Q3 - Q1
    upper = Q3 + 1.5 * IQR

    return group > upper

cust_df['is_outlier'] = cust_df.groupby('product_category')['purchase_amount'].transform(flag_outliers)
outliers = cust_df[cust_df['is_outlier'] == True]
print(f'Outliers when checked WITHIN each category: {len(outliers)}')
print()
print(outliers[['order_id', 'product_category', 'purchase_amount']].sort_values('purchase_amount', ascending=False).head(10).to_string())
```

#### We are going to cap our outliers instead of deleting them.

```python
cust_df = pd.read_csv('step4_invalid.csv')

def cap_outliers(group):
    Q1, Q3 = group.quantile(0.25), group.quantile(0.75)
    IQR = Q3 - Q1
    upper = Q3 + 1.5 * IQR
    return group.clip(upper=upper) # clip(upper=upper) caps any value above the bound to go down

cust_df['purchase_amount'] = cust_df.groupby('product_category')['purchase_amount'].transform(cap_outliers)

print('New max purchase_amount per category:')
print(cust_df.groupby('product_category')['purchase_amount'].max())

cust_df.to_csv('step5_outliers.csv', index=False)


 
```

#### Inputation/ handling missing values

```python
cust_df = pd.read_csv('step5_outliers.csv')

#Numeric: median imputation
cust_df['age'] = cust_df['age'].fillna(cust_df['age'].median())

#Numeric, category-aware: group median
cust_df['purchase_amount'] = cust_df.groupby('product_category')['purchase_amount'].transform(
    lambda x: x.fillna(x.median())
)

#categorical: 'unknown' rather than guessing, avoid skewing the data
cust_df['gender'] = cust_df['gender'].fillna('Unknown')
cust_df['city'] = cust_df['city'].fillna('Unknown')

#domain-reasoned fills (Not statistical imputation, but bussiness logic)
cust_df['rating'] = cust_df['rating'].fillna('Not rated')
cust_df['membership_status'] = cust_df['membership_status'].fillna('None')

print('remaining missing values:')
print(cust_df.isnull().sum().sum(), 'Total nulls left')
print()
print(cust_df.isnull().sum())

cust_df.to_csv('step6_imputed.csv', index=False)
```
